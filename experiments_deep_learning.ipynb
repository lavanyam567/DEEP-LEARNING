```python

# EXPERIMENT 1: VECTOR AND MATRIX OPERATIONS FOR DEEP LEARNING

# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics.pairwise import cosine_similarity

# Step 1: Create Customer-Genre Matrix (6x5)
customer_genre = np.array([
    [5, 2, 4, 1, 3],
    [4, 2, 5, 1, 3],
    [1, 5, 2, 4, 3],
    [2, 4, 1, 5, 3],
    [5, 3, 4, 2, 1],
    [1, 4, 2, 3, 5]
])

# Step 2: Create Genre-Book Matrix (5x4)
genre_book = np.array([
    [0.9, 0.2, 0.5, 0.1],
    [0.3, 0.8, 0.2, 0.6],
    [0.7, 0.4, 0.9, 0.2],
    [0.1, 0.9, 0.3, 0.8],
    [0.4, 0.3, 0.6, 0.9]
])

print("Customer-Genre Matrix:")
print(customer_genre)

print("\nGenre-Book Matrix:")
print(genre_book)

# Step 3: Calculate Dot Product
dot_product = np.dot(customer_genre[0], genre_book[:, 0])

print("\nDot Product:")
print(dot_product)

# Step 4: Matrix Multiplication
score_matrix = np.matmul(customer_genre, genre_book)

print("\nCustomer-Book Score Matrix:")
print(score_matrix)

# Step 5: Calculate Transpose
print("\nTranspose of Customer-Genre Matrix:")
print(customer_genre.T)

# Step 6: Calculate Inverse of Valid Square Matrix
square_matrix = customer_genre.T @ customer_genre

print("\nSquare Matrix:")
print(square_matrix)

print("\nInverse of Square Matrix:")
print(np.linalg.inv(square_matrix))

# Step 7: Element-wise Addition
print("\nElement-wise Addition:")
print(customer_genre + customer_genre)

# Step 8: Element-wise Multiplication
print("\nElement-wise Multiplication:")
print(customer_genre * customer_genre)

# Step 9: Calculate Cosine Similarity
similarity = cosine_similarity(customer_genre)

print("\nCosine Similarity Matrix:")
print(np.round(similarity, 3))

# Identify two most similar customers
similarity_without_diagonal = similarity.copy()
np.fill_diagonal(similarity_without_diagonal, 0)

i, j = np.unravel_index(
    np.argmax(similarity_without_diagonal),
    similarity_without_diagonal.shape
)

print("\nMost Similar Customers:")
print("Customer", i + 1, "and Customer", j + 1)
print("Similarity Score:", round(similarity[i, j], 3))

# Step 10: Implement Dense Layer y = Wx + b
np.random.seed(42)

x = customer_genre[0]
W = np.random.rand(4, 5)
b = np.random.rand(4)

y = np.dot(W, x) + b

print("\nDense Layer Calculation:")
print("Input Vector:", x)
print("Weight Matrix:\n", W)
print("Bias:", b)
print("Output:", y)

# Step 11: Visualise Customer-Book Recommendation Scores
plt.figure(figsize=(9, 5))

sns.heatmap(
    score_matrix,
    annot=True,
    cmap="YlGnBu",
    fmt=".1f",
    xticklabels=["Book 1", "Book 2", "Book 3", "Book 4"],
    yticklabels=[
        "Customer 1", "Customer 2", "Customer 3",
        "Customer 4", "Customer 5", "Customer 6"
    ]
)

plt.title("Customer-Book Recommendation Score Matrix")
plt.xlabel("Books")
plt.ylabel("Customers")
plt.show()

# Step 12: Visualise Cosine Similarity
plt.figure(figsize=(7, 5))

sns.heatmap(
    similarity,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Customer Cosine Similarity Heatmap")
plt.xlabel("Customers")
plt.ylabel("Customers")
plt.show()

print("\nExperiment 1 Completed Successfully!")
```

```python

# EXPERIMENT 2: COVARIANCE, EIGENVALUES AND EIGENVECTORS ANALYSIS

# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Generate synthetic patient dataset
np.random.seed(42)

n = 200

hemoglobin = np.random.normal(13.5, 1.5, n)
rbc = hemoglobin * 0.35 + np.random.normal(3.5, 0.5, n)
glucose = np.random.normal(110, 20, n)
cholesterol = glucose * 1.2 + np.random.normal(100, 15, n)
bmi = np.random.normal(25, 4, n)

data = pd.DataFrame({
    'Hemoglobin': hemoglobin,
    'RBC': rbc,
    'Glucose': glucose,
    'Cholesterol': cholesterol,
    'BMI': bmi
})

print("Patient Dataset:")
print(data.head())

print("\nDataset Shape:", data.shape)

# Step 2: Standardise the features
standardized_data = (data - data.mean()) / data.std()

print("\nStandardised Data:")
print(standardized_data.head())

# Step 3: Calculate covariance matrix
cov_matrix = np.cov(standardized_data.T)

print("\nCovariance Matrix:")
print(cov_matrix)

# Step 4: Calculate correlation matrix
corr_matrix = standardized_data.corr()

print("\nCorrelation Matrix:")
print(corr_matrix)

# Step 5: Plot covariance and correlation heatmaps
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.heatmap(
    cov_matrix,
    annot=True,
    cmap='coolwarm',
    xticklabels=data.columns,
    yticklabels=data.columns
)
plt.title("Covariance Matrix")

plt.subplot(1, 2, 2)
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='viridis',
    xticklabels=data.columns,
    yticklabels=data.columns
)
plt.title("Correlation Matrix")

plt.tight_layout()
plt.show()

# Step 6: Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

# Sort eigenvalues in descending order
indices = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[indices]
eigenvectors = eigenvectors[:, indices]

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)

# Step 7: Calculate percentage of variance
explained_variance = (eigenvalues / np.sum(eigenvalues)) * 100

print("\nPercentage of Variance Captured:")
for i, value in enumerate(explained_variance):
    print(f"Principal Component {i+1}: {value:.2f}%")

# Step 8: Plot explained variance
plt.figure(figsize=(8, 5))

plt.bar(
    range(1, 6),
    explained_variance,
    color='skyblue',
    label='Individual Variance'
)

plt.plot(
    range(1, 6),
    np.cumsum(explained_variance),
    marker='o',
    color='red',
    label='Cumulative Variance'
)

plt.xlabel("Principal Components")
plt.ylabel("Variance Percentage")
plt.title("Explained Variance by Eigenvalues")
plt.xticks(range(1, 6))
plt.legend()
plt.grid()
plt.show()

# Step 9: Plot patient data with top two eigenvectors
plt.figure(figsize=(8, 6))

x = standardized_data.iloc[:, 0]
y = standardized_data.iloc[:, 1]

plt.scatter(x, y, alpha=0.5, label="Patient Data")

scale = 3

for i in range(2):
    plt.arrow(
        0, 0,
        eigenvectors[0, i] * scale,
        eigenvectors[1, i] * scale,
        color=['red', 'green'][i],
        width=0.015,
        head_width=0.12,
        label=f"Eigenvector {i+1}"
    )

plt.xlabel("Standardised Hemoglobin")
plt.ylabel("Standardised RBC")
plt.title("Patient Data with Top Two Eigenvector Directions")
plt.legend()
plt.grid()
plt.show()

print("\nExperiment 2 Completed Successfully!")
```
